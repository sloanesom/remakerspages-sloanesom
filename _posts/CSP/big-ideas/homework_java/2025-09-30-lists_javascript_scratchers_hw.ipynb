### Hack 1 – Add Up Numbers

Make a list of numbers. Write code to:
1. Find the total sum.
2. Find the average.

```python
%%js

let numArray = [1, 2, 3, 2, 4, 2, 5];
let targetNumber = 2;

// Write your code here:
let count = numArray.reduce((accumulator, currentValue) => {
  if (currentValue === targetNumber) {
    return accumulator + 1;
  } else {
    return accumulator;
  }
}, 0);

console.log(count); // Output: 3

```

### Hack 2 – Count Repeats

Make a list with repeated items. Write code to count how many times each item appears.

```python
%%js

import java.util.ArrayList;
import java.util.HashMap;

public class CountItems {
    public static void main(String[] args) {
        // Make a list with repeated items
        ArrayList<String> fruits = new ArrayList<String>();
        fruits.add("apple");
        fruits.add("banana");
        fruits.add("apple");
        fruits.add("cherry");
        fruits.add("banana");
        fruits.add("apple");

        // Create a HashMap to store counts
        HashMap<String, Integer> counts = new HashMap<String, Integer>();

        // Loop through each fruit
        for (String fruit : fruits) {
            // If it's already in the map, add 1 to its count
            if (counts.containsKey(fruit)) {
                counts.put(fruit, counts.get(fruit) + 1);
            } 
            // If it's not in the map, start its count at 1
            else {
                counts.put(fruit, 1);
            }
        }

        // Print results
        System.out.println("Item counts:");
        for (String fruit : counts.keySet()) {
            System.out.println(fruit + ": " + counts.get(fruit));
        }
    }
}
```

### Hack 3 – Keep Only Evens

Make a list of numbers. Write code to create a new list with only even numbers.

```python
%%js

import java.util.ArrayList;

public class EvenNumbers {
    public static void main(String[] args) {
        // Make a list of numbers
        ArrayList<Integer> numbers = new ArrayList<Integer>();
        for (int i = 1; i <= 10; i++) {
            numbers.add(i);
        }

        // Create a new list for even numbers
        ArrayList<Integer> evenNumbers = new ArrayList<Integer>();

        // Loop through and pick out the evens
        for (int num : numbers) {
            if (num % 2 == 0) {
                evenNumbers.add(num);
            }
        }

        // Print the result
        System.out.println("Even numbers: " + evenNumbers);
    }
}
```
