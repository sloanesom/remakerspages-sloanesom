### Hack 1 – Add Up Numbers

Make a list of numbers. Write code to:
1. Find the total sum.
2. Find the average.

```python
// Hack 1 – Add Up Numbers
let numbers = [4, 7, 1, 9, 6, 7, 10];

// Write your code here:
let sum = numbers.reduce((accumulator, currentValue) => accumulator + currentValue, 0);

console.log(sum); // Output: 44

// Hack 2 – Find the Largest Number
let moreNumbers = [3, 5, 2, 8, 1, 4];

// Write your code here:
let largest = moreNumbers.reduce((max, current) => (current > max ? current : max), moreNumbers[0]);

console.log(largest); // Output: 8

// Hack 3 – Count Occurrences of a Specific Number
let numArray = [1, 2, 3, 2, 4, 2, 5];
let targetNumber = 2;
```
